## Plot 1

**Scenario**: A researcher wants to show that Gentoo penguins are significantly heavier than other species, but their plot is visually overwhelming and exaggerates the differences.

```python
penguins = sns.load_dataset("penguins")

ax = sns.barplot(data=penguins, x="species", y="body_mass_g", palette=["blue", "orange", "green"],
    legend=False)
ax.set_ylim(0, 5500) 
plt.title("Penguin Mass by Species")
plt.show()
```

Issues with the plot: The Range of the y axis started at 3000, which made it more favor the Gentoo race. The colors or unprofessional and are not needed. 


How you changed the plot and why: Making it start at 0 creates a more fair range. Next, we changed the colors since they are not professional. We changed them to blue, orange, and green.


## Plot 2

**Scenario**: A student plotted the carat size versus price for 54,000 diamonds to see where the bulk of the market sits, but the result is just a solid block of ink.

```python
diamonds["carat_range"] = pd.cut(
    diamonds["carat"],
    bins=[0, 0.5, 1, 1.5, 2, 3, 6],
    labels=["0–0.5", "0.5–1", "1–1.5", "1.5–2", "2–3", "3–6"]
)

plt.figure(figsize=(10, 6))
sns.boxplot(
    data=diamonds,
    x="carat_range",
    y="price",
    color="lightblue",
    showfliers=False
)

plt.title("Diamond Prices by Carat Range")
plt.xlabel("Diamond Weight (carats)")
plt.ylabel("Price (USD)")
plt.show()
```

Issues with the plot: Cluttered, ugly, confusing, axis titles not clear/capitalized


How you changed the plot and why: We decided that the scatter plot was inefficient and cluttered so we decided box plot better. Then we changed the axis to giver a more clear description. As well as the intervals for the diamond weight. This shows the median as well as the amount in each range.


## Plot 3

The aviation board wants to see passenger growth month-over-month for each year in the 1950s, but the lines are indistinguishable.

```python
flights = sns.load_dataset("flights")

flights_table = flights.pivot(
    index="year",
    columns="month",
    values="passengers"
)

sns.heatmap(flights_table, cmap="YlOrRd")

plt.title("Monthly Airline Passengers by Year")
plt.xlabel("Month")
plt.ylabel("Year")
plt.show()
```

Issues with the plot: Every year is shown with the same black line, so it is almost impossible to tell which line represents which year. There is also no legend identifying the years. At first, I tried separating them by color but it still looked really bad and really hard to see which year, month and how many passengers.

How you changed the plot and why: I changed the line graph to a heatmap because it makes it easier to compare passenger numbers across both months and years. The lighter and darker colors show lower and higher passenger numbers, so I can see the overall growth over the years and the changes throughout each year more clearly. As you can see in the graph when there is the less number of passengers it is yellow and as they are more passengers it gets more reddish, the middle point is like an orange color.

## Plot 4

**Scenario**: A restaurant manager receives this chart showing the relationship between total bills and tips, split by smoking preference. 





```python
tips = sns.load_dataset("tips")

sns.lmplot(data=tips, x="total_bill", y="tip", hue="smoker", scatter_kws={"alpha": 0.5})
plt.title("Total Bill vs. Tip by Smoker Status")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

Issues with the plot: The plot is a little difficult to understand at first because it does not have a title and the axis labels are not very clear. Some of the points also overlap, which makes it harder to see all of the observations.


How you changed the plot and why: I added a title and clearer axis labels so it is easier to understand what the graph is showing and that the bill and tip are in dollars. I also made the points more transparent so it is easier to see when observations overlap. I kept the two trend lines because they make it easier to compare the relationship between the total bill and tip for smokers and non-smokers.
